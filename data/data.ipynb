```python
import pandas as pd
import pyarrow.parquet as pq
from IPython.display import display

parquet_path = "DeepScaler/Qwen2d5_math_7b/train_800.success_rate_k8.right.parquet"
```

```python
# 1. 读取数据
df = pd.read_parquet(parquet_path)

# 2. 查看总数据量
print(f"=== 总数据量 ===")
print(f"共 {len(df)} 条数据\n")
```

```python
# 3. 查看字段列表
print(f"=== 字段列表 ({len(df.columns)} 个) ===")
print(df.columns.tolist(), "\n")
```

```python
# 4. 查看第一条示例数据 (转换为字典方便打印)
print(f"=== 第 1 条数据示例 ===")
first_row = df.iloc[0].to_dict()
for key, value in first_row.items():
    print(f"[{key}]: {value}")
```

```python
# 统计 difficulty_bucket 各类别的数量
counts = df['difficulty_bucket'].value_counts()

print("=== 难度分布统计 (数量) ===")
print(counts)

print("\n=== 难度分布统计 (占比) ===")
print(df['difficulty_bucket'].value_counts(normalize=True) * 100)
```

```python
parquet_file = pq.ParquetFile(parquet_path)

print(f"行数: {parquet_file.metadata.num_rows}")
print(f"列数: {parquet_file.metadata.num_columns}")
print(f"Row Groups 数量: {parquet_file.num_row_groups}")

print("\n字段结构：")
print(parquet_file.schema_arrow)
```

```python
schema_df = pd.DataFrame([
    {
        "column_name": field.name,
        "arrow_type": str(field.type),
        "nullable": field.nullable,
    }
    for field in parquet_file.schema_arrow
])

display(schema_df)
```
